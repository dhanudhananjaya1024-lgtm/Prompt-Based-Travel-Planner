```python
from groq import Groq
import os

client = Groq(
    api_key=os.environ["GROQ_API_KEY"]
)

MODEL = "llama-3.3-70b-versatile"

print("Travel Planner AI is ready!")
```

```python
TRAVEL_PROMPT = """
You are an expert AI Travel Planner.

Create a practical travel plan based on the user's requirements.

USER REQUIREMENTS
-----------------
Destination: {destination}
Budget: {budget}
Number of Days: {days}
Interests: {interests}
Number of Travelers: {travelers}

Generate the response using the following sections:

1. TRIP SUMMARY
Give a short overview of the trip.

2. DAY-BY-DAY ITINERARY
Create a detailed itinerary for every day.
Include:
- Morning activity
- Afternoon activity
- Evening activity
- Approximate activity cost

3. HOTEL SUGGESTIONS
Suggest 3 types of accommodation:
- Budget
- Mid-range
- Premium

Give approximate prices and explain why each type is suitable.

4. PACKING LIST
Create a useful packing checklist based on:
- Destination
- Number of days
- Interests
- General travel needs

5. COST ESTIMATE
Provide an estimated breakdown for:
- Transportation
- Hotel
- Food
- Activities
- Miscellaneous

Calculate an approximate total cost.

6. TRAVEL TIPS
Give useful safety, transportation and money-saving tips.

Important:
- Keep the plan realistic.
- Stay within the user's budget as much as possible.
- Clearly mention when costs are approximate.
- Do not claim that prices or hotel availability are real-time.
- Use Indian Rupees (INR) unless another currency is specified.
- Return only the travel plan.
"""
```

```python
def create_travel_plan(
    destination,
    budget,
    days,
    interests,
    travelers
):

    prompt = TRAVEL_PROMPT.format(
        destination=destination,
        budget=budget,
        days=days,
        interests=interests,
        travelers=travelers
    )

    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {
                "role": "system",
                "content": "You are an expert travel planning assistant."
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0.6,
        max_completion_tokens=5000
    )

    return response.choices[0].message.content
```

```python
def generate_plan(destination, budget, days, interests, travelers):

    if not destination.strip():
        return "❌ Please enter a destination."

    if not budget.strip():
        return "❌ Please enter your budget."

    if not days.strip():
        return "❌ Please enter the number of days."

    if not interests.strip():
        return "❌ Please enter your interests."

    if not travelers.strip():
        travelers = "1"

    try:
        result = create_travel_plan(
            destination,
            budget,
            days,
            interests,
            travelers
        )

        return result

    except Exception as e:
        return f"❌ Error: {str(e)}"
```

```python
import gradio as gr

with gr.Blocks(
    title="AI Travel Planner"
) as app:

    gr.Markdown(
        """
        # ✈️ AI Travel Planner

        ### Prompt-Based Personalized Travel Planning

        Enter your destination, budget, travel duration and interests.
        The AI will create a complete travel plan.
        """
    )

    # ---------------------------
    # INPUT SECTION
    # ---------------------------

    gr.Markdown("## 📝 Trip Details")

    with gr.Row():

        with gr.Column():

            destination = gr.Textbox(
                label="📍 Destination",
                placeholder="Example: Mysore, Karnataka"
            )

            budget = gr.Textbox(
                label="💰 Budget",
                placeholder="Example: ₹20,000"
            )

            days = gr.Textbox(
                label="📅 Number of Days",
                placeholder="Example: 3"
            )

        with gr.Column():

            interests = gr.Textbox(
                label="❤️ Interests",
                placeholder="Example: History, Food, Nature, Photography",
                lines=4
            )

            travelers = gr.Textbox(
                label="👥 Number of Travelers",
                placeholder="Example: 2",
                value="1"
            )

    generate_button = gr.Button(
        "✨ Generate Travel Plan",
        variant="primary"
    )

    # ---------------------------
    # OUTPUT SECTION
    # ---------------------------

    gr.Markdown("## 🗺️ Your AI Travel Plan")

    output = gr.Markdown()

    generate_button.click(
        fn=generate_plan,
        inputs=[
            destination,
            budget,
            days,
            interests,
            travelers
        ],
        outputs=output
    )

    gr.Markdown(
        """
        ---
        ### Project Information

        **Project:** Prompt-Based Travel Planner

        **Technologies:** Python, Groq LLM, Gradio

        **Concepts:** Prompt Engineering, Prompt Templates,
        Step-by-Step Task Decomposition
        """
    )

app.launch(share=True)
```
